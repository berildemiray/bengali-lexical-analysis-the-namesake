# Findings and Linguistic Interpretation

The corpus analysis demonstrates that Bengali lexical elements appear consistently throughout The Namesake and function as important markers of cultural continuity, multilingual identity and diasporic belonging. The extracted lexical items are recurring multilingual elements embedded within English discourse across the novel. The frequency analysis reveals that culturally marked lexical items occur far more frequently than expected in a predominantly English-language text. Terms such as sari, baba and dal appear repeatedly throughout the corpus, indicating the persistent presence of Bengali cultural identity within everyday communication. The recurrence of these terms suggests that multilingual lexical practices remain central to the characters’ social and cultural environment despite geographical displacement.

The lexical item “Bengali” appears with the highest frequency within the corpus, which suggests that linguistic and cultural identity constitute one of the central thematic concerns of the novel and reinforces the persistent emphasis on cultural belonging throughout the narrative. Furthermore, kinship terminology constitutes one of the most prominent lexical categories within the corpus. Words such as baba appear frequently in family-related interactions and are closely associated with intimacy, familiarity and interpersonal belonging. Their repeated occurrence indicates that family discourse functions as one of the primary spaces in which Bengali linguistic identity is preserved within the novel. The corpus also contains a high number of culturally specific lexical items connected to clothing, food, rituals and domestic life. Lexical elements such as sari and dal preserve cultural references that remain embedded within English narration without being fully translated. The continued use of these expressions suggests that the narrative intentionally maintains elements of Bengali linguistic and cultural specificity rather than assimilating completely into monolingual English discourse. Likewise, geographical references such as Calcutta and Howrah appear repeatedly throughout the corpus and function as markers of spatial and cultural attachment. Their recurrence reinforces the continued symbolic presence of the homeland within the characters’ daily lives and highlights the importance of geographical memory within diasporic discourse.

The analysis further demonstrates that most multilingual insertions occur at lexical level rather than sentence level. The majority of the corpus remains structurally English while selectively incorporating Bengali lexical items into dialogue and narration. This pattern suggests that multilingual identity in the novel is expressed primarily through culturally marked lexical choices embedded within dominant English syntactic structures. Another significant finding is that multilingual lexical items frequently occur in emotionally significant or culturally intimate contexts, particularly during family interactions, celebrations and discussions related to tradition or belonging. The placement of Bengali lexical elements within these contexts indicates that multilingual language practices in the novel are strongly connected to emotional expression, cultural intimacy and identity preservation. The frequency distribution additionally shows that some lexical categories appear substantially more often than others. Cultural objects, kinship terminology and place names dominate the corpus, whereas larger sentence-level multilingual structures occur less frequently. This suggests that the novel prioritizes culturally symbolic lexical insertions over extensive bilingual dialogue, creating a hybrid linguistic environment while maintaining overall readability for English-language audiences.

Overall, the corpus findings demonstrate that multilingual lexical elements in The Namesake function as persistent markers of cultural identity, diasporic belonging and linguistic hybridity. The analysis further shows how corpus-based computational methods such as tokenization and lexical frequency analysis can reveal meaningful multilingual patterns within literary texts and support linguistic interpretation through quantitative textual analysis.
